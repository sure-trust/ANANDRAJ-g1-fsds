# Tag removal


```python
import re
```

```python
df['cleaned']=df['text'].apply(lambda x:str(x).lower())
```

```python
df.head()
```

```python
def rem_tag(text):
    cleaned_text = re.sub(re.compile('<*?.>'),'',str(text))

    return cleaned_text
```

```python
df['cleaned'] = df['cleaned'].apply(rem_tag)
```

```python
df.head()
```

```python
def rem_char(text):
    text = re.sub(re.compile('<a-zA-Z0-9\s>'),'',str(text))
    text = re.sub(r'\s',' ',str(text)).strip()

    return text
```

```python
df['cleaned'] = df['cleaned'].apply(rem_char)

df.head()
```

```python
def normalize(text):
    if isinstance(text,str):
        text = re.sub(r'[^\w\s]',"",text)
        text = re.sub(r"\s+",' ', text).strip()
    else:
        text = str(text)

    return text

df['cleaned'] = df['cleaned'].apply(normalize)

```

```python
df.head()
```

```python
import nltk
nltk.download('stopwords')
```

```python
stop_words = stopwords.words('english')
df['cleaned'] = df['cleaned'].apply(lambda x:[i for i in x.split() if i not in stop_words]).apply(lambda x:" ".join(x))
```

```python
df.head()
```

```python
def remove_urls(text):
    pattern = re.compile(r'https?://\S+|www\.\S+')
    pattern = re.compile(r'http?://\S+|www\.\S+')
    return pattern.sub(r'',text)
```

```python
df['cleaned'] = df['text'].apply(remove_urls)
```

```python
df['cleaned'] = df['cleaned'].apply(lambda x:[i for i in x.split() if i not in stop_words]).apply(lambda x:" ".join(x))
df['cleaned'] = df['cleaned'].apply(normalize)
df['cleaned'] = df['cleaned'].apply(rem_char)
df['cleaned'] = df['cleaned'].apply(rem_tag)
```

```python
df['cleaned']=df['cleaned'].apply(lambda x:str(x).lower())
```

```python
df.head()
```

```python
import string

a = string.punctuation

def remove_punctuation(text):
    return text.translate(str.maketrans('','',a))
```

```python
df['cleaned'] = df['cleaned'].apply(remove_punctuation)
```

```python
df.head()
```

```python
X= df['cleaned']
y = df['sentiment']

```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()
y = encoder.fit_transform(y)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=13)
```

```python
cv = CountVectorizer()
```

```python
  X_train_processed= cv.fit_transform(X_train).toarray()
  X_test_processed = cv.transform(X_test).toarray()
```

# Model building

```python
from sklearn.ensemble import RandomForestClassifier


rf = RandomForestClassifier()
rf.fit(X_train_processed,y_train)
```

```python
print("Training Data Shape:", X_train_processed.shape)
print("Test Data Shape:", X_test_processed.shape)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

y_pred = rf.predict(X_test_processed)

# performance
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
```
